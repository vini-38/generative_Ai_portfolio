```python
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Flatten
from tensorflow .keras.utils import to_categorical

(x_train,y_train),(x_test,y_test)=mnist.load_data()
print(x_train.shape)
print(x_test.shape)
plt.imshow(x_train[0],cmap="gray")
plt.title("input image")
plt.axis("off")
plt.show()

x_train=x_train/255.0
x_test=x_test/255.0
y_train=to_categorical(y_train,10)
y_test=to_categorical(y_test,10)
model=Sequential([Flatten(input_shape=(28,28)),
                  Dense(128,activation="relu"),
                  Dense(64,activation="relu"),
                  Dense(10,activation="softmax")])
model.compile(optimizer="adam",
              loss="categorical_crossentropy",
              metrics=["accuracy"])
model.summary()


model.fit(x_train,y_train,epochs=5,batch_size=32,validation_split=0.1)
test_loss,test_accuracy=model.evaluate(x_test,y_test)
print(test_accuracy)
prediction=model.predict(x_test[0:1])
predicted_digit=np.argmax(prediction)
print("actual_digit")
print("predicted_digit")
plt.imshow(x_test[0],cmap="gray")
plt.title("output")
plt.axis("off")
plt.show()

```

```python
#multiple input and differnt lenth
import numpy as np

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow .keras.layers import Embedding,SimpleRNN,Dense,Masking
sentence=["I Love Machine",
          "I Love Deep Learning",
          "I really love data Science"]
output_word=["learning","models","analytics"]
tokenizer=Tokenizer()
tokenizer.fit_on_texts(sentence+output_word)
x=tokenizer.texts_to_sequences(sentence)
x=pad_sequences(x,maxlen=5,padding="post")
print("after padding",x)
y=np.array([
    tokenizer.word_index["learning"],
    tokenizer.word_index["models"],
    tokenizer.word_index["analytics"]
])
model=Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1,output_dim=16),
    Masking(mask_value=0),
    SimpleRNN(32),
    Dense(len(tokenizer.word_index)+1,activation="softmax")])
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy")
model.fit(x,y,epochs=500,verbose=0)
prediction=model.predict(x)
predicted_ids=np.argmax(prediction,axis=1)

#convert index into word
index_to_word={value:key for key,value in tokenizer.word_index.items()}
for i in range(len(predicted_ids)):
 print(sentence[i],"->",index_to_word.get(predicted_ids[i]))

```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding,Dense,LSTM
from tensorflow.keras.models import Sequential
import numpy as np
sentence=["i love this movie","this movie is excellent","i hate this movie","i dislike the movie"]
label=np.array([1,1,0,0])
tokenizer=Tokenizer(num_words=500)
tokenizer.fit_on_texts(sentence)
print(tokenizer.word_index)
sequences=tokenizer.texts_to_sequences(sentence)
print(sequences)
x=pad_sequences(sequences,maxlen=5,padding="post")
print(x)
model=Sequential([Embedding(input_dim=500,output_dim=10,input_length=5),LSTM(32),Dense(1,activation="sigmoid")])
model.compile(optimizer="adam",loss="binary_crossentropy")
model.fit(x,label,epochs=100,batch_size=2)
new_sent=["i love this wonderful movie"]
new_sequences=tokenizer.texts_to_sequences(new_sent)
new_pad=pad_sequences(new_sequences,maxlen=5,padding="post")
prediction=model.predict(new_pad)
print(prediction)
if prediction >=0.5:
  print("positive")
else:
  print("negative")
```

```python
#cnn
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import matplotlib.pyplot as plt
import numpy as np

(x_train,y_train),(x_test,y_test)=tf.keras.datasets.mnist.load_data()
print(y_train.shape)
print(x_train.shape)
print(y_test.shape)
print(x_test.shape)

x_train=x_train/255.0
x_test=x_test/255.0
x_train=x_train.reshape(-1,28,28,1)
x_test=x_test.reshape(-1,28,28,1)
model=Sequential()
model.add(Conv2D(32,(3,3),activation="relu",input_shape=(28,28,1)))
model.add(MaxPooling2D(2,2))
model.add(Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2))
model.add(Flatten())
model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation="softmax"))
model.summary()

model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
model.fit(x_train,y_train,epochs=3,validation_split=0.1)
test_loss,test_accuracy=model.evaluate(x_test,y_test)
print(test_loss)
print(test_accuracy)
index=10
image=x_test[index]
actual_label=y_test[index]

plt.imshow(image.reshape(28,28),cmap="grey")
plt.title("actual_label")
plt.axis("off")
plt.show()

image_prediction=image.reshape(1,28,28,1)
prediction=model.predict(image_prediction)
predicted_digit=np.argmax(prediction)
print(actual_label)
print(predicted_digit)

```

```python
# Data agumentation(image turn panna)

import tensorflow as tf
from tensorflow.keras import layers,models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import matplotlib.pyplot as plt

(x_train,y_train),(x_test,y_test)=tf.keras.datasets.mnist.load_data()
x_train=x_train.reshape(-1,28,28,1)
x_test=x_test.reshape(-1,28,28,1)

x_train=x_train/255.0
x_test=x_test/255.0

datagen=ImageDataGenerator(rotation_range=15,width_shift_range=0.1,height_shift_range=0.1,zoom_range=0.1)

text_generator=datagen.flow(x_train,y_train,batch_size=64)


model=models.Sequential([
    layers.Conv2D(32,(3,3),activation="relu",input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64,(3,3),activation="relu"),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),

    layers.Dense(64,activation="relu"),
    layers.Dense(10,activation="softmax"),


])

model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])

model.fit(text_generator,epochs=5,validation_data=(x_test,y_test))
test_loss,test_accuracy=model.evaluate(x_test,y_test)
print(test_loss)
print(test_accuracy)

index=8

image=x_test[index]
actual_label=y_test[index]
print(actual_label)
# for actual image
plt.imshow(image.reshape(28,28),cmap="grey")
plt.title("actual image")
plt.axis("off")
plt.show()

input_image=np.expand_dims(image,axis=0)

prediction=model.predict(input_image)

predicted_label=np.argmax(prediction)
print(predicted_label)

# for predicted image
plt.imshow(image.reshape(28,28),cmap="grey")
plt.title("predicted image")
plt.axis("off")
plt.show()






```
